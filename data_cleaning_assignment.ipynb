# Data Cleaning and Exploration using Pandas

Data Engineering Internship Assignment

```
import pandas as pd
import numpy as np
```

## Load Dataset

```
df = pd.read_csv("Combined_dataset.csv")
df.head()
```

## Explore the Dataset

```
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Info:")
df.info()

print("\nLast 5 Rows:")
df.tail()
```

## Handle Missing Values

```
print(df.isnull().sum())

text_columns = [
    "what_customers_said",
    "variations",
    "seller_information",
    "seller_name"
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].fillna("Not Available")

if "discount" in df.columns:
    df["discount"] = df["discount"].fillna(df["discount"].mean())

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

## Basic Data Operations

```
selected_columns = df[["title", "rating", "initial_price"]]
selected_columns.head()

high_rating = df[df["rating"] > 4]
high_rating.head()
```

## Remove Duplicates

```
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after removal:", df.duplicated().sum())
```

## Create Derived Columns

```
df["quantity"] = np.random.randint(1, 5, size=len(df))

df["total_amount"] = df["initial_price"] * df["quantity"]

df[["initial_price", "quantity", "total_amount"]].head()
```

## Save Cleaned Dataset

```
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Summary

- Loaded the CSV dataset using Pandas.
- Explored the dataset using `head()`, `tail()`, `shape`, `columns`, and `info()`.
- Handled missing values by filling text fields with `"Not Available"` and numeric values with the column mean.
- Performed basic data operations such as filtering rows and selecting columns.
- Removed duplicate rows.
- Created the derived columns `quantity` and `total_amount`.
- Exported the cleaned dataset as `cleaned_dataset.csv`.
